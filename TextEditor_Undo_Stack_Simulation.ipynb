```python
class Node:
  def __init__(self, text) -> None:
    self.text = text
    self.next = None
```

```python
class undoSimulation:
  def __init__(self) -> None:
    self.top = None

  def pushWord(self, text):
    newNode = Node(text)
    if self.top is None:
      self.top = newNode
      return
    else:
      newNode.next = self.top
      self.top = newNode

  def popWord(self):
    if self.top is None:
      return
    else:
      self.top = self.top.next

  def showWords(self):
    if self.top is None:
      print("")
      return

    words = []
    current = self.top
    while current is not None:
      words.append(current.text)
      current = current.next

    for word in reversed(words):
      print(f" {word}", end=" ")
    print()
```

```python
def main():
  us = undoSimulation()
  us.pushWord("Hello")
  us.showWords()
  us.pushWord("World")
  us.showWords()
  us.popWord()
  us.showWords()
  us.popWord()
  us.showWords()
```

```python
if __name__ == "__main__":
  main()
```
